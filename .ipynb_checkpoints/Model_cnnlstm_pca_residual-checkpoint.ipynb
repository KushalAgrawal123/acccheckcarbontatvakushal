```python
"""
CarbonTatva AI — Model: CNN-LSTM + PCA + Residual Analysis
===========================================================
Market : GDAM only
Data   : Last-year (Mar 2025 – Feb 2026)
Split  : Train ≤ Sep-2025  |  Val ≤ Nov-2025  |  Test > Nov-2025

Changes vs previous version:
  1. MARKETS = ['GDAM'] only
  2. PCA applied per feature-group after StandardScaler:
       - PAST_FEATURES  → pca_past  (fed into CNN-LSTM encoder)
       - FUTURE_FEATURES → pca_future (fed into future projection head)
     Each PCA fitted only on train, then applied to val/test.
     Variance threshold 0.995 for both groups.
  3. Residual analysis block added (same diagnostic suite as other models):
     - hourly bias, price-bucket bias, weekday bias, autocorrelation
"""

import os, pickle, warnings
import numpy as np
import pandas as pd
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

warnings.filterwarnings('ignore')
OUT = 'outputs_cnnlstm_pca'
os.makedirs(OUT, exist_ok=True)

# ─────────────────────────────────────────────────────────────────
# DEVICE
# ─────────────────────────────────────────────────────────────────
def get_device():
    if torch.cuda.is_available():
        device = torch.device('cuda')
        print(f"  Device: CUDA ({torch.cuda.get_device_name(0)})")
    elif hasattr(torch.backends, 'mps') and torch.backends.mps.is_available():
        device = torch.device('mps')
        print("  Device: MPS (Apple Silicon GPU)")
    else:
        device = torch.device('cpu')
        print("  Device: CPU")
    return device

# ─────────────────────────────────────────────────────────────────
# CONFIG
# ─────────────────────────────────────────────────────────────────
DATA_FILE       = 'final_merged_data.xlsx'
MARKETS         = ['GDAM']
SLOTS_DAY       = 96
LAST_YEAR_START = '2025-03-01'
TRAIN_END       = '2025-09-30'
VAL_END         = '2025-11-30'
LOOKBACK        = SLOTS_DAY * 2
HORIZON         = SLOTS_DAY
PCA_VAR         = 0.995
SEED            = 42
BATCH_SIZE      = 64
EPOCHS          = 80

torch.manual_seed(SEED)
np.random.seed(SEED)

# ─────────────────────────────────────────────────────────────────
# FEATURE ENGINEERING
# ─────────────────────────────────────────────────────────────────
def build_features(df):
    df = df.copy()
    d = pd.to_datetime(df['Delivery Date'], dayfirst=True)
    t = df['Time Period'].str.split('-').str[0]
    df['datetime'] = pd.to_datetime(d.astype(str) + ' ' + t)
    df = df.sort_values('datetime').reset_index(drop=True)

    df.drop(columns=['apparent_temperature (°C)',
                     'sunshine_duration (s)',
                     'precipitation (mm)'], errors='ignore', inplace=True)

    df['hour']  = df['datetime'].dt.hour
    df['dow']   = df['datetime'].dt.dayofweek
    df['month'] = df['datetime'].dt.month

    df['hour_sin']  = np.sin(2 * np.pi * df['hour']  / 24)
    df['hour_cos']  = np.cos(2 * np.pi * df['hour']  / 24)
    df['month_sin'] = np.sin(2 * np.pi * df['month'] / 12)
    df['month_cos'] = np.cos(2 * np.pi * df['month'] / 12)
    df['dow_sin']   = np.sin(2 * np.pi * df['dow']   / 7)
    df['dow_cos']   = np.cos(2 * np.pi * df['dow']   / 7)
    df['is_weekend']   = (df['dow'] >= 5).astype(int)
    df['is_evening']   = ((df['hour'] >= 18) & (df['hour'] <= 21)).astype(int)
    df['is_solar_hrs'] = ((df['hour'] >= 8)  & (df['hour'] <= 16)).astype(int)

    try:
        import holidays
        india_h = holidays.India(years=range(2022, 2027))
        df['holiday_flag'] = df['datetime'].dt.date.apply(
            lambda d: 1 if d in india_h else 0)
    except ImportError:
        df['holiday_flag'] = 0

    p = df['Price (Rs./MWh)']
    df['price_lag96']  = p.shift(96)
    df['price_lag672'] = p.shift(672)
    df['ma3d'] = p.shift(1).rolling(288).mean()
    df['ma7d'] = p.shift(1).rolling(672).mean()
    df['price_diff1d'] = p.shift(96)  - p.shift(192)
    df['price_diff1w'] = p.shift(672) - p.shift(1344)
    df['rel_lag96']  = df['price_lag96']  / (df['ma7d'] + 1e-6)
    df['rel_lag672'] = df['price_lag672'] / (df['ma7d'] + 1e-6)

    df['temp']    = df['temperature_2m (°C)']
    df['temp_sq'] = df['temp'] ** 2
    df['radiation']            = df['shortwave_radiation (W/m²)']
    df['radiation_x_hour_sin'] = df['radiation'] * df['hour_sin']
    df['wind_speed']  = df['wind_speed_100m (km/h)']
    df['cloud_cover'] = df['cloud_cover (%)']
    df['humidity']    = df['relative_humidity_2m (%)']
    df['rain']        = df['rain (mm)']

    df['log_price'] = np.log1p(p)
    return df


PAST_FEATURES = [
    'log_price',
    'hour_sin', 'hour_cos',
    'month_sin', 'month_cos',
    'dow_sin', 'dow_cos',
    'is_weekend', 'is_evening',
    'radiation', 'temp', 'wind_speed', 'cloud_cover',
    'rel_lag96',
]

FUTURE_FEATURES = [
    'hour_sin', 'hour_cos',
    'month_sin', 'month_cos',
    'dow_sin', 'dow_cos',
    'is_weekend', 'holiday_flag',
    'is_evening', 'is_solar_hrs',
    'price_lag96', 'price_lag672',
    'rel_lag96', 'rel_lag672',
    'ma3d', 'ma7d',
    'price_diff1d', 'price_diff1w',
    'radiation', 'radiation_x_hour_sin',
    'temp', 'temp_sq',
    'wind_speed', 'cloud_cover', 'humidity',
]

# ─────────────────────────────────────────────────────────────────
# DATASET  — uses pre-PCA-transformed arrays directly
# ─────────────────────────────────────────────────────────────────
class ElectricityDataset(Dataset):
    """
    past_arr  : (N, n_past_pca)  — already scaled + PCA-transformed
    future_arr: (N, n_future_pca) — already scaled + PCA-transformed
    log_prices: (N,)
    """
    def __init__(self, past_arr, future_arr, log_prices,
                 lookback=LOOKBACK, horizon=HORIZON):
        self.past   = past_arr.astype(np.float32)
        self.future = future_arr.astype(np.float32)
        self.lp     = log_prices.astype(np.float32)
        self.lookback = lookback
        self.horizon  = horizon

    def __len__(self):
        return len(self.lp) - self.lookback - self.horizon + 1

    def __getitem__(self, idx):
        p = idx + self.lookback
        return (torch.from_numpy(self.past[idx:p]),
                torch.from_numpy(self.future[p:p+self.horizon]),
                torch.from_numpy(self.lp[p:p+self.horizon]))


# ─────────────────────────────────────────────────────────────────
# MODEL
# ─────────────────────────────────────────────────────────────────
class AttentionLayer(nn.Module):
    def __init__(self, hidden_size):
        super().__init__()
        self.attn = nn.Linear(hidden_size, 1)

    def forward(self, x):
        w = torch.softmax(self.attn(x).squeeze(-1), dim=1)
        return (x * w.unsqueeze(-1)).sum(dim=1)


class CNNLSTM(nn.Module):
    def __init__(self, n_past, n_future,
                 cnn_filters=48, lstm_hidden=96,
                 lstm_layers=2, horizon=HORIZON, dropout=0.3):
        super().__init__()
        self.horizon = horizon
        self.cnn3 = nn.Sequential(
            nn.Conv1d(n_past, cnn_filters, kernel_size=3, padding=1),
            nn.BatchNorm1d(cnn_filters), nn.GELU())
        self.cnn5 = nn.Sequential(
            nn.Conv1d(n_past, cnn_filters, kernel_size=5, padding=2),
            nn.BatchNorm1d(cnn_filters), nn.GELU())
        self.cnn7 = nn.Sequential(
            nn.Conv1d(n_past, cnn_filters, kernel_size=7, padding=3),
            nn.BatchNorm1d(cnn_filters), nn.GELU())

        cnn_out = cnn_filters * 3
        self.lstm = nn.LSTM(cnn_out, lstm_hidden, lstm_layers,
                            batch_first=True, bidirectional=True,
                            dropout=dropout if lstm_layers > 1 else 0)
        lstm_out = lstm_hidden * 2

        self.attention   = AttentionLayer(lstm_out)
        self.future_proj = nn.Sequential(
            nn.Linear(n_future, 32), nn.LayerNorm(32), nn.GELU())

        fusion_in = lstm_out + 32 * horizon
        self.output_net = nn.Sequential(
            nn.Linear(fusion_in, 192), nn.LayerNorm(192),
            nn.GELU(), nn.Dropout(dropout),
            nn.Linear(192, 96), nn.GELU(),
            nn.Dropout(dropout / 2),
            nn.Linear(96, horizon))

    def forward(self, xp, xf):
        B = xp.shape[0]
        c = xp.permute(0, 2, 1)
        c = torch.cat([self.cnn3(c), self.cnn5(c), self.cnn7(c)], dim=1).permute(0, 2, 1)
        c = self.attention(self.lstm(c)[0])
        f = self.future_proj(xf).reshape(B, -1)
        return self.output_net(torch.cat([c, f], dim=1))


# ─────────────────────────────────────────────────────────────────
# UTILITIES
# ─────────────────────────────────────────────────────────────────
def compute_metrics(y_true, y_pred, label=''):
    mae  = np.mean(np.abs(y_true - y_pred))
    rmse = np.sqrt(np.mean((y_true - y_pred) ** 2))
    mask = y_true > 0
    mape = np.mean(np.abs((y_true[mask] - y_pred[mask]) / y_true[mask])) * 100
    print(f"    {label:20s}  MAE={mae:7.1f}  RMSE={rmse:7.1f}  MAPE={mape:.2f}%")
    return {'mae': mae, 'rmse': rmse, 'mape': mape}


def eval_loader(model, loader, device):
    model.eval()
    preds, targets = [], []
    with torch.no_grad():
        for xp, xf, y in loader:
            xp, xf = xp.to(device), xf.to(device)
            preds.append(model(xp, xf).cpu().numpy())
            targets.append(y.numpy())
    p_price = np.clip(np.expm1(np.concatenate(preds).flatten()),   0, 20000)
    t_price = np.clip(np.expm1(np.concatenate(targets).flatten()), 0, 20000)
    return t_price, p_price


def save_pred_plot(y_true, y_pred, title, path):
    fig, axes = plt.subplots(3, 1, figsize=(16, 12))
    fig.patch.set_facecolor('#0f1117')
    for ax in axes:
        ax.set_facecolor('#1a1d27'); ax.tick_params(colors='#aaa')
    n = min(800, len(y_true))
    axes[0].plot(y_true[:n], color='#aaa', lw=0.8, label='Actual')
    axes[0].plot(y_pred[:n], color='#2980b9', lw=0.8, label='Predicted')
    axes[0].set_title(f'{title} — Test Set', color='white', fontsize=12)
    axes[0].legend(fontsize=9); axes[0].set_ylabel('Price (Rs./MWh)', color='#aaa')
    n2 = SLOTS_DAY * 3
    axes[1].plot(y_true[:n2], color='#aaa', lw=1.5, label='Actual')
    axes[1].plot(y_pred[:n2], color='#2980b9', lw=1.5, ls='--', label='Predicted')
    axes[1].set_title(f'{title} — First 3 Days', color='white', fontsize=12)
    axes[1].legend(fontsize=9); axes[1].set_ylabel('Price (Rs./MWh)', color='#aaa')
    axes[2].scatter(y_true[:3000], y_pred[:3000], s=2, alpha=0.3, color='#2980b9')
    lim = max(y_true[:3000].max(), y_pred[:3000].max())
    axes[2].plot([0, lim], [0, lim], 'r--', lw=1.5, label='Perfect')
    axes[2].set_title(f'{title} — Actual vs Predicted', color='white', fontsize=12)
    axes[2].set_xlabel('Actual', color='#aaa'); axes[2].set_ylabel('Predicted', color='#aaa')
    axes[2].legend(fontsize=9)
    plt.tight_layout()
    plt.savefig(path, dpi=150, bbox_inches='tight', facecolor='#0f1117')
    plt.close()


# ─────────────────────────────────────────────────────────────────
# RESIDUAL ANALYSIS
# ─────────────────────────────────────────────────────────────────
def residual_analysis(test_df, y_true, y_pred, mkt, out_dir):
    """
    y_true / y_pred are flattened price arrays aligned with test_df rows
    (HORIZON-step-ahead predictions rolled out over test set).
    We align by trimming test_df to match the predictions length.
    """
    n = len(y_pred)
    df = test_df.iloc[:n].copy().reset_index(drop=True)
    df['residual'] = y_pred - y_true[:n]
    df['abs_err']  = np.abs(df['residual'])
    df['hour']     = df['datetime'].dt.hour
    df['dow']      = df['datetime'].dt.dayofweek
    df['date']     = df['datetime'].dt.date

    yt = y_true[:n]
    q33, q67    = np.percentile(yt, 33), np.percentile(yt, 67)
    spike_thr   = np.percentile(yt, 95)
    def price_bucket(p):
        if p < q33:         return '① Low (<P33)'
        elif p < q67:       return '② Mid (P33-P67)'
        elif p < spike_thr: return '③ High (P67-P95)'
        else:               return '④ Spike (>P95)'
    df['bucket'] = yt
    df['bucket'] = df['bucket'].apply(price_bucket)

    day_names = ['Mon','Tue','Wed','Thu','Fri','Sat','Sun']
    bkt_order = ['① Low (<P33)', '② Mid (P33-P67)', '③ High (P67-P95)', '④ Spike (>P95)']

    fig, axes = plt.subplots(2, 2, figsize=(16, 12))
    fig.patch.set_facecolor('#0f1117')
    fig.suptitle(f'Residual Analysis — CNN-LSTM+PCA GDAM\n'
                 f'(+ve = overestimate, –ve = underestimate)', color='white', fontsize=13)
    for ax in axes.flat:
        ax.set_facecolor('#1a1d27'); ax.tick_params(colors='#aaa')
        ax.axhline(0, color='white', lw=0.8, ls='--', alpha=0.5)

    hrly = df.groupby('hour')['residual'].mean()
    colors_hr = ['#e74c3c' if v > 0 else '#2980b9' for v in hrly.values]
    axes[0,0].bar(hrly.index, hrly.values, color=colors_hr, alpha=0.85)
    axes[0,0].set_title('Bias by Hour of Day', color='white')
    axes[0,0].set_xlabel('Hour', color='#aaa')
    axes[0,0].set_ylabel('Mean Residual (Rs./MWh)', color='#aaa')
    axes[0,0].set_xticks(range(0, 24, 2))
    for h in hrly.abs().nlargest(3).index:
        axes[0,0].annotate(f'{hrly[h]:+.0f}', xy=(h, hrly[h]),
                           xytext=(h, hrly[h] + (30 if hrly[h]>0 else -30)),
                           color='yellow', fontsize=8, ha='center')

    bkt_bias = df.groupby('bucket')['residual'].mean().reindex(bkt_order)
    bkt_mae  = df.groupby('bucket')['abs_err'].mean().reindex(bkt_order)
    x = np.arange(len(bkt_order)); w = 0.4
    colors_bkt = ['#e74c3c' if v > 0 else '#2980b9' for v in bkt_bias.values]
    axes[0,1].bar(x - w/2, bkt_bias.values, w, color=colors_bkt, alpha=0.85, label='Bias')
    axes[0,1].bar(x + w/2, bkt_mae.values,  w, color='#f39c12',  alpha=0.7,  label='MAE')
    axes[0,1].set_title('Bias & MAE by Price Regime', color='white')
    axes[0,1].set_xticks(x)
    axes[0,1].set_xticklabels([b.split('(')[0].strip() for b in bkt_order],
                               rotation=15, ha='right', fontsize=9)
    axes[0,1].set_ylabel('Rs./MWh', color='#aaa'); axes[0,1].legend(fontsize=9)
    for i, (b, m) in enumerate(zip(bkt_bias.values, bkt_mae.values)):
        axes[0,1].text(i-w/2, b + (15 if b>=0 else -25), f'{b:+.0f}', ha='center',
                       color='white', fontsize=8)
        axes[0,1].text(i+w/2, m+5, f'{m:.0f}', ha='center', color='white', fontsize=8)

    dow_bias = df.groupby('dow')['residual'].mean()
    dow_mae  = df.groupby('dow')['abs_err'].mean()
    x3 = np.arange(7)
    colors_dow = ['#e74c3c' if v > 0 else '#2980b9' for v in dow_bias.values]
    axes[1,0].bar(x3 - w/2, dow_bias.values, w, color=colors_dow, alpha=0.85, label='Bias')
    axes[1,0].bar(x3 + w/2, dow_mae.values,  w, color='#f39c12',  alpha=0.7,  label='MAE')
    axes[1,0].set_title('Bias & MAE by Weekday', color='white')
    axes[1,0].set_xticks(x3); axes[1,0].set_xticklabels(day_names)
    axes[1,0].set_ylabel('Rs./MWh', color='#aaa'); axes[1,0].legend(fontsize=9)
    axes[1,0].axvspan(4.5, 6.5, color='white', alpha=0.05)
    axes[1,0].text(5.5, axes[1,0].get_ylim()[1]*0.9, 'Weekend',
                   color='#aaa', ha='center', fontsize=8)

    res = df['residual'].values
    max_lag = 48
    acf_vals = [np.corrcoef(res[:-lag], res[lag:])[0,1] for lag in range(1, max_lag+1)]
    ci = 1.96 / np.sqrt(len(res))
    colors_acf = ['#e74c3c' if abs(v) > ci else '#555' for v in acf_vals]
    axes[1,1].bar(range(1, max_lag+1), acf_vals, color=colors_acf, alpha=0.85)
    axes[1,1].axhline( ci, color='yellow', lw=1, ls='--', label=f'95% CI (±{ci:.3f})')
    axes[1,1].axhline(-ci, color='yellow', lw=1, ls='--')
    axes[1,1].set_title('Residual Autocorrelation (Lags 1–48 slots)', color='white')
    axes[1,1].set_xlabel('Lag (15-min slots)', color='#aaa')
    axes[1,1].set_ylabel('Correlation', color='#aaa')
    axes[1,1].legend(fontsize=9)
    sig_lags = [i+1 for i, v in enumerate(acf_vals) if abs(v) > ci]
    if sig_lags:
        axes[1,1].text(max_lag*0.6, 0.8*max(acf_vals),
                       f'Significant: {sig_lags[:5]}{"…" if len(sig_lags)>5 else ""}',
                       color='yellow', fontsize=8)

    plt.tight_layout()
    plt.savefig(f'{out_dir}/residual_analysis_{mkt}.png',
                dpi=150, bbox_inches='tight', facecolor='#0f1117')
    plt.close()
    print(f"  Saved: {out_dir}/residual_analysis_{mkt}.png")

    day_mae = df.groupby('date')['abs_err'].mean().nlargest(10)
    print(f"\n  Top-10 worst predicted days (by MAE):")
    for dt, mae_val in day_mae.items():
        sub = df[df['date'] == dt]
        bias_val = sub['residual'].mean()
        direction = 'OVER' if bias_val > 0 else 'UNDER'
        print(f"    {dt}  MAE={mae_val:.0f}  avg_bias={bias_val:+.0f} ({direction})")

    print(f"\n  ── Residual Summary ──────────────────────────────────")
    print(f"  Overall bias (mean residual): {df['residual'].mean():+.1f} Rs./MWh")
    print(f"  Most overestimated hour : {hrly.idxmax():02d}:00  ({hrly.max():+.0f})")
    print(f"  Most underestimated hour: {hrly.idxmin():02d}:00  ({hrly.min():+.0f})")
    print(f"  Highest MAE price regime: {bkt_mae.idxmax()}  (MAE={bkt_mae.max():.0f})")
    sig_pct = 100 * len(sig_lags) / max_lag
    print(f"  Autocorrelated lags     : {len(sig_lags)}/{max_lag} ({sig_pct:.0f}%) — "
          + ("serial structure remains" if len(sig_lags) > 5 else "mostly white noise ✓"))


# ─────────────────────────────────────────────────────────────────
# MAIN
# ─────────────────────────────────────────────────────────────────
print("=" * 60)
print("MODEL: CNN-LSTM + PCA (Last Year Data)")
print(f"Market: GDAM only")
print(f"Period: {LAST_YEAR_START} → Feb 2026")
print(f"Split:  Train≤{TRAIN_END}  |  Val≤{VAL_END}  |  Test>Val")
print(f"LOOKBACK={LOOKBACK} slots ({LOOKBACK//96} days), HORIZON={HORIZON}")
print(f"PCA variance threshold: {PCA_VAR} (applied per feature group)")
print("=" * 60)
DEVICE = get_device()

sheets    = pd.read_excel(DATA_FILE, sheet_name=None)
all_feats = list(dict.fromkeys(PAST_FEATURES + FUTURE_FEATURES))

for mkt in MARKETS:
    print(f"\n{'─'*50}\nMarket: {mkt}\n{'─'*50}")

    df_raw   = build_features(sheets[mkt])
    df_clean = df_raw.dropna(subset=all_feats + ['log_price']).copy()
    df_clean = df_clean[df_clean['datetime'] >= LAST_YEAR_START].copy()
    print(f"  Rows: {len(df_clean):,}  ({df_clean['datetime'].min().date()} → {df_clean['datetime'].max().date()})")

    train_df = df_clean[df_clean['datetime'] <= TRAIN_END]
    val_df   = df_clean[(df_clean['datetime'] > TRAIN_END) & (df_clean['datetime'] <= VAL_END)]
    test_df  = df_clean[df_clean['datetime'] > VAL_END]
    print(f"  Train: {len(train_df):,} | Val: {len(val_df):,} | Test: {len(test_df):,}")

    # ── StandardScaler per feature group ─────────────────────────
    scaler_past   = StandardScaler()
    scaler_future = StandardScaler()

    # Fit on train only; log_price stays unscaled
    tr_past_s  = scaler_past.fit_transform(train_df[PAST_FEATURES].values)
    va_past_s  = scaler_past.transform(val_df[PAST_FEATURES].values)
    te_past_s  = scaler_past.transform(test_df[PAST_FEATURES].values)

    tr_fut_s   = scaler_future.fit_transform(train_df[FUTURE_FEATURES].values)
    va_fut_s   = scaler_future.transform(val_df[FUTURE_FEATURES].values)
    te_fut_s   = scaler_future.transform(test_df[FUTURE_FEATURES].values)

    # ── PCA per feature group ─────────────────────────────────────
    pca_past   = PCA(n_components=PCA_VAR, random_state=SEED)
    pca_future = PCA(n_components=PCA_VAR, random_state=SEED)

    tr_past_p  = pca_past.fit_transform(tr_past_s)
    va_past_p  = pca_past.transform(va_past_s)
    te_past_p  = pca_past.transform(te_past_s)

    tr_fut_p   = pca_future.fit_transform(tr_fut_s)
    va_fut_p   = pca_future.transform(va_fut_s)
    te_fut_p   = pca_future.transform(te_fut_s)

    n_past_p   = pca_past.n_components_
    n_future_p = pca_future.n_components_
    print(f"\n  PCA past  : {len(PAST_FEATURES)} → {n_past_p} components "
          f"({pca_past.explained_variance_ratio_.cumsum()[-1]*100:.2f}%)")
    print(f"  PCA future: {len(FUTURE_FEATURES)} → {n_future_p} components "
          f"({pca_future.explained_variance_ratio_.cumsum()[-1]*100:.2f}%)")

    # PCA variance plots
    for pca_obj, label, color in [(pca_past, 'Past', '#2980b9'),
                                   (pca_future, 'Future', '#27ae60')]:
        n_c = pca_obj.n_components_
        fig, ax = plt.subplots(figsize=(10, 4))
        fig.patch.set_facecolor('#0f1117'); ax.set_facecolor('#1a1d27')
        ax.bar(range(1, n_c+1), pca_obj.explained_variance_ratio_ * 100,
               color=color, alpha=0.85)
        ax.plot(range(1, n_c+1), pca_obj.explained_variance_ratio_.cumsum() * 100,
                color='#f39c12', lw=2, marker='o', ms=3, label='Cumulative')
        ax.axhline(99.5, color='yellow', lw=1, ls='--', label='99.5%')
        ax.set_title(f'PCA ({label} features) — {mkt}', color='white')
        ax.set_xlabel('Component', color='#aaa')
        ax.set_ylabel('Variance Explained (%)', color='#aaa')
        ax.legend(fontsize=9); ax.tick_params(colors='#aaa')
        plt.tight_layout()
        plt.savefig(f'{OUT}/pca_variance_{label.lower()}_{mkt}.png',
                    dpi=150, bbox_inches='tight', facecolor='#0f1117')
        plt.close()

    # ── Build datasets ────────────────────────────────────────────
    tr_lp = train_df['log_price'].values
    va_lp = val_df['log_price'].values
    te_lp = test_df['log_price'].values

    train_ds = ElectricityDataset(tr_past_p, tr_fut_p, tr_lp)
    val_ds   = ElectricityDataset(va_past_p, va_fut_p, va_lp)
    test_ds  = ElectricityDataset(te_past_p, te_fut_p, te_lp)
    print(f"  Samples — Train: {len(train_ds):,} | Val: {len(val_ds):,} | Test: {len(test_ds):,}")

    train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True,  num_workers=0)
    val_loader   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False, num_workers=0)
    test_loader  = DataLoader(test_ds,  batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

    # ── Model ─────────────────────────────────────────────────────
    model = CNNLSTM(n_past_p, n_future_p, 48, 96, 2, HORIZON, 0.3).to(DEVICE)
    n_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"  Parameters: {n_params:,}  (input dims: past={n_past_p}, future={n_future_p})")

    optimizer = torch.optim.AdamW(model.parameters(), lr=5e-4, weight_decay=1e-3)

    def lr_lambda(epoch):
        WARMUP = 5
        if epoch < WARMUP: return epoch / WARMUP
        return 0.5 * (1 + np.cos(np.pi * (epoch - WARMUP) / max(EPOCHS - WARMUP, 1)))

    scheduler = torch.optim.lr_scheduler.LambdaLR(optimizer, lr_lambda)
    criterion = nn.HuberLoss(delta=1.5)

    PATIENCE   = 12
    best_val   = float('inf')
    pat_count  = 0
    train_losses, val_losses = [], []

    print(f"  Training CNN-LSTM on PCA features (max {EPOCHS} epochs)...")

    for epoch in range(1, EPOCHS + 1):
        model.train()
        ep_loss = 0.0
        for xp, xf, y in train_loader:
            xp, xf, y = xp.to(DEVICE), xf.to(DEVICE), y.to(DEVICE)
            optimizer.zero_grad()
            loss = criterion(model(xp, xf), y)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()
            ep_loss += loss.item()

        model.eval()
        vl = 0.0
        with torch.no_grad():
            for xp, xf, y in val_loader:
                xp, xf, y = xp.to(DEVICE), xf.to(DEVICE), y.to(DEVICE)
                vl += criterion(model(xp, xf), y).item()

        tl = ep_loss / len(train_loader)
        vl = vl / len(val_loader)
        train_losses.append(tl); val_losses.append(vl)
        scheduler.step()

        lr_now = optimizer.param_groups[0]['lr']
        print(f"    Epoch {epoch:3d}/{EPOCHS}  train={tl:.4f}  val={vl:.4f}  lr={lr_now:.6f}")

        if vl < best_val:
            best_val = vl; pat_count = 0
            torch.save(model.state_dict(), f'{OUT}/cnnlstm_{mkt}_best.pt')
        else:
            pat_count += 1
            if pat_count >= PATIENCE:
                print(f"    Early stopping at epoch {epoch}")
                break

    model.load_state_dict(torch.load(f'{OUT}/cnnlstm_{mkt}_best.pt', map_location=DEVICE))

    y_val_true,  y_val_pred  = eval_loader(model, val_loader,  DEVICE)
    y_test_true, y_test_pred = eval_loader(model, test_loader, DEVICE)

    print("  Results:")
    m_val  = compute_metrics(y_val_true,  y_val_pred,  'Validation')
    m_test = compute_metrics(y_test_true, y_test_pred, 'Test')

    # Training curve
    fig, ax = plt.subplots(figsize=(10, 4))
    fig.patch.set_facecolor('#0f1117'); ax.set_facecolor('#1a1d27')
    ax.plot(train_losses, color='#e74c3c', lw=1.2, label='Train Loss')
    ax.plot(val_losses,   color='#27ae60', lw=1.2, label='Val Loss')
    ax.axvline(int(np.argmin(val_losses)), color='yellow', lw=1.5, ls='--', label='Best epoch')
    ax.set_title(f'CNN-LSTM+PCA Curve — {mkt}', color='white')
    ax.set_xlabel('Epoch', color='#aaa'); ax.set_ylabel('Huber Loss', color='#aaa')
    ax.legend(fontsize=9); ax.tick_params(colors='#aaa')
    plt.tight_layout()
    plt.savefig(f'{OUT}/cnnlstm_loss_{mkt}.png', dpi=150, bbox_inches='tight', facecolor='#0f1117')
    plt.close()

    save_pred_plot(y_test_true, y_test_pred,
                   f'CNN-LSTM+PCA — {mkt}',
                   f'{OUT}/cnnlstm_pred_{mkt}.png')

    # ── RESIDUAL ANALYSIS ─────────────────────────────────────────
    print(f"\n  Running residual analysis...")
    residual_analysis(test_df, y_test_true, y_test_pred, mkt, OUT)

    with open(f'{OUT}/cnnlstm_pipeline_{mkt}.pkl', 'wb') as f:
        pickle.dump({
            'scaler_past': scaler_past, 'scaler_future': scaler_future,
            'pca_past': pca_past, 'pca_future': pca_future,
            'past_features': PAST_FEATURES, 'future_features': FUTURE_FEATURES,
            'n_past_pca': n_past_p, 'n_future_pca': n_future_p,
            'lookback': LOOKBACK, 'horizon': HORIZON,
        }, f)
    print(f"  Saved: {OUT}/cnnlstm_{mkt}_best.pt + pipeline pkl")

print(f"\n{'='*60}")
print(f"Done! Check {OUT}/")
print(f"Device used: {DEVICE}")
```
